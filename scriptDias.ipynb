```python
import pandas as pd

# Cargar datos
df = pd.read_excel('DF_CLIMA.xlsx', engine='openpyxl')
cal = pd.read_excel('calendario_enero2020-diciembre2025.xlsx', engine='openpyxl')

# Normalizar columnas del calendario
cal.columns = cal.columns.str.lower()

# Convertir columna fecha
cal['dia'] = pd.to_datetime(cal['dia'], dayfirst=True)

# Extraer fecha sin hora en DF final
df['timestamp'] = pd.to_datetime(df['timestamp'], dayfirst=True)
df['fecha_solo'] = df['timestamp'].dt.date
cal['fecha_solo'] = cal['dia'].dt.date

# Unir calendario al DF principal
df = df.merge(cal[['fecha_solo', 'dia_semana', 'laborable / festivo / domingo festivo']], on='fecha_solo', how='left')

# Crear columna festivo (si / no)
df['festivo'] = df['laborable / festivo / domingo festivo'].str.contains('festivo', case=False, na=False).map({True:'si', False:'no'})

# Crear columna dias_habiles (si entre lunes-viernes, no en sábados/domingo)
dias_no_habiles = ['sabado', 'sábado', 'domingo']
df['dias_habiles'] = df['dia_semana'].str.lower().isin(dias_no_habiles).map({True:'no', False:'si'})

# Guardar resultado final en Excel
df.to_excel('DF_FINAL.xlsx', index=False)

df.head()
```
