```python
import pandas as pd

print("=== Cargando datos ===")
df_trafico = pd.read_excel("DF_FINAL_CONCLIMA.xlsx", engine="openpyxl")
df_clima = pd.read_excel("AEMET_Madrid_Retiro_2021_2025_DatosProcesados.xlsx", engine="openpyxl")

print(f"Filas DF_FINAL: {len(df_trafico)}")
print(f"Filas AEMET: {len(df_clima)}")

# --- Normalizar tiempo ---
df_trafico["timestamp"] = pd.to_datetime(df_trafico["timestamp"], errors="coerce")
df_trafico = df_trafico.sort_values("timestamp").reset_index(drop=True)

# --- Calcular horas por día ---
horas_por_dia = int(round(df_trafico["timestamp"].dt.date.value_counts().mean()))
print(f"Horas promedio por día: {horas_por_dia}")

# --- Repetir clima ---
df_clima_expanded = df_clima.loc[df_clima.index.repeat(horas_por_dia)].reset_index(drop=True)

# --- Igualar longitud ---
n = min(len(df_trafico), len(df_clima_expanded))
df_trafico = df_trafico.iloc[:n].reset_index(drop=True)
df_clima_expanded = df_clima_expanded.iloc[:n].reset_index(drop=True)

# --- Unir ---
df_final = pd.concat([df_trafico, df_clima_expanded], axis=1)

# --- Guardar en CSV plano ---
salida_csv = "DF_FINAL_CONCLIMA_COMPLETO.csv"
df_final.to_csv(salida_csv, index=False, encoding="utf-8")

print(f"✅ Archivo CSV generado correctamente: {salida_csv}")
print(f"Filas: {len(df_final)} | Columnas: {len(df_final.columns)}")
```
