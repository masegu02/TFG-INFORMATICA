```python
import pandas as pd
import glob, os

carpeta = "."
patron = os.path.join(carpeta, "procesado_aemet_3195_Madrid_Retiro_*.csv")
archivos = sorted(glob.glob(patron))
if not archivos:
    raise FileNotFoundError("No hay CSV 'procesado_aemet_3195_Madrid_Retiro_*.csv'.")

dfs = []
for f in archivos:
    # 1) Detecta separador ; o ,
    try:
        df = pd.read_csv(f, sep=";", engine="python")
        if df.shape[1] == 1:  # cayó todo en una sola columna → reintenta con coma
            df = pd.read_csv(f, sep=",", engine="python")
    except Exception:
        df = pd.read_csv(f, sep=",", engine="python")

    # 2) Normaliza cabeceras y limpia
    df.columns = [c.strip() for c in df.columns]
    if "fecha" not in df.columns:
        raise ValueError(f"'{os.path.basename(f)}' no tiene columna 'fecha'.")

    df = df.dropna(subset=["fecha"])
    df["fecha"] = df["fecha"].astype(str).str.strip()

    # 3) Parseo robusto de fechas (acepta dd/mm/yyyy y yyyy-mm-dd[ HH:MM:SS])
    dt = pd.to_datetime(df["fecha"], errors="coerce", dayfirst=True, infer_datetime_format=True)
    mask_bad = dt.isna()
    if mask_bad.any():
        # segundo intento quitando hora si viniera pegada
        df_loc = df.loc[mask_bad, "fecha"].str.slice(0, 10)
        dt2 = pd.to_datetime(df_loc, errors="coerce", dayfirst=True)
        dt.loc[mask_bad] = dt2

    df["fecha"] = dt
    df = df.dropna(subset=["fecha"])

    # 4) Guarda origen y añade
    df["origen_archivo"] = os.path.basename(f)
    print(f"{os.path.basename(f)} -> filas válidas: {len(df)}")
    dfs.append(df)

# 5) Unir, ordenar y deduplicar por fecha
df_final = pd.concat(dfs, ignore_index=True)
df_final = df_final.sort_values("fecha").drop_duplicates(subset=["fecha"]).reset_index(drop=True)

# 6) Exportar
df_final.to_excel("AEMET_Madrid_Retiro_2021_2025.xlsx", index=False)
print(f"OK -> AEMET_Madrid_Retiro_2021_2025.xlsx  Filas: {len(df_final)}  Rango: {df_final['fecha'].min()} .. {df_final['fecha'].max()}")
```
