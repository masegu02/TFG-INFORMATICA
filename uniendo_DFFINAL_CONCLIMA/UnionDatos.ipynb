```python
import pandas as pd

print("=== INICIO ===")

# 1) Cargar
df = pd.read_excel("DF_FINAL_CONCLIMA.xlsx")
aemet = pd.read_excel("AEMET_Madrid_Retiro_2021_2025_DatosProcesados.xlsx")
print(f"DF: {df.shape}  AEMET: {aemet.shape}")

# 2) Seleccionar columnas AEMET (incluye precipitación)
cols_aemet = ["tmed","tmin","tmax","prec","presMax","presMin","hrMedia","hrMax","hrMin"]
presentes = [c for c in cols_aemet if c in aemet.columns]
aemet = aemet[presentes]
print("AEMET columnas usadas:", presentes)

# 3) Repetir cada fila de AEMET 24 veces y ajustar a DF
aemet_exp = aemet.loc[aemet.index.repeat(24)].reset_index(drop=True)
aemet_exp = aemet_exp.iloc[:len(df)]
if len(aemet_exp) < len(df):
    df = df.iloc[:len(aemet_exp)]
print(f"Expandido AEMET: {aemet_exp.shape}  DF ajustado: {df.shape}")

# 4) Renombrar con prefijo
aemet_exp.columns = [f"AEMET_{c}" for c in aemet_exp.columns]

# 5) Unir por columnas
df_final = pd.concat([df.reset_index(drop=True), aemet_exp], axis=1)
print("Combinado:", df_final.shape)

# 6) Guardar CSV legible por Excel (separador ;)
df_final.to_csv("DF_FINAL_CONCLIMA_COMBINADO.csv", index=False, sep=";", encoding="utf-8-sig")
print("OK -> DF_FINAL_CONCLIMA_COMBINADO.csv")
```
